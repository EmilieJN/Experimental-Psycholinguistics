```python
import pandas as pd #short form for pandas for ease. pd is the standard for pandas.
```

```python
a = [3, 4, 5, 6]
b = [7, 8, 9, 10]
c = [11, 12, 13, 14]
#each list needs to have the same amount of items. this is a requirement of pandas'; it likes squares.
```

```python
abc = pd.DataFrame({'a':a, 'b':b, 'c':c}) 
#looks like a dictionary based on syntax. ends up creating a structured overview over the data in form of a table.
```

```python
type(abc)
```

```python
abc
```

```python
#to get data out again. shows the content of 'a' along with their numbering.
abc['a']
```

```python
d = abc['a']
type(d)
```

```python
e = list(d) #converting into list to see the numbers of a plainly, like OG a.
e
```

```python
a == e #== means compare these two items
```
