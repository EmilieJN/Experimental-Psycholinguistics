### Pandas Cheat Sheet <br>

Use this cheat sheet if you get stuck with common pandas functions, methods, or syntax. <br>

### Looking at your dataframe <br>

`df.columns` returns the names of all columns in the dataframe. <br>

`df.head()` shows the first 5 rows of the dataframe. <br>

`df.shape` returns the number of rows and columns in the dataframe as `(rows, columns)`. <br>

### Checking for missing values <br>

`df['columnname'].notna()` returns `True` for rows where the value is not missing (`NaN`). <br>

For example: `condition = df['correctResponse'].notna()` <br>

You can then use this condition to select those rows: `df_clean = df.loc[condition, :]` <br>

`df['columnname'].isna()` returns `True` for rows where the value is missing (`NaN`). <br>

### Selecting columns <br>

`df['columnname']` selects one column. <br>

`df.loc[:, 'columnname']` also selects one column. The `:` means all rows. <br>

To select multiple columns, put the column names inside a list: `df.loc[:, ['column1', 'column2', 'column3']]` <br>

For example: `df.loc[:, ['choice', 'correctResponse', 'duration']]` <br>

### Selecting rows <br>

`df.loc[df['columnname'] == 'value', :]` selects rows according to a condition. The `:` means all columns. <br>

For example: `df.loc[df['correct'] == True, :]` selects all rows where `correct` is `True`. <br>

The basic structure of `.loc[]` is `df.loc[rows, columns]`. <br>

You can select particular rows and columns at the same time: `df.loc[df['correct'] == True, ['choice', 'duration']]` <br>

This selects rows where `correct` is `True`, but only keeps the `choice` and `duration` columns. <br>

### Using multiple conditions <br>

Put each condition inside parentheses and connect them using `&`. <br>

For example: `df.loc[(df['correct'] == True) & (df['duration'] < 2000), :]` <br>

This selects rows where `correct` is `True` AND `duration` is below `2000`. <br>

You can add more conditions in the same way: `df.loc[(df['correct'] == True) & (df['duration'] < 2000) & (df['condition'] == 'experimental'), :]` <br>

You can also select only certain columns: `df.loc[(df['correct'] == True) & (df['duration'] < 2000), ['choice', 'duration']]` <br>

Use `|` instead of `&` if you want OR instead. <br>

### Checking whether values are in a list <br>

`df['columnname'].isin(listname)` checks whether each value in a column appears in a list. <br>

For example: `vowels = ['a', 'e', 'i', 'o', 'u']` <br>

`df['choice'].isin(vowels)` returns `True` when the value in `choice` is found in `vowels`. <br>

You can combine `.isin()` with `.loc[]`: `df.loc[df['choice'].isin(vowels), 'letter_type'] = 'vowel'` <br>

### Changing values based on a condition <br>

`df.loc[condition, 'columnname'] = 'new value'` puts a value in a column for rows where the condition is `True`. <br>

For example: `df.loc[df['duration'] > 2000, 'speed'] = 'slow'` <br>

This puts `'slow'` in the `speed` column whenever `duration` is above `2000`. <br>

You can also use multiple conditions: `df.loc[(df['duration'] > 2000) & (df['correct'] == True), 'speed'] = 'slow but correct'` <br>

### Moving a column <br>

`column = df.pop('columnname')` removes the specified column from the dataframe and saves its values in `column`. <br>

You can then insert it somewhere else using `df.insert(where, 'columnname', column)`. <br>

For example: `df.insert(2, 'columnname', column)` inserts the column as the third column. <br>

Remember that Python starts counting at `0`: `0` is the first column, `1` is the second column, and `2` is the third column. <br>

### Grouping data with `.groupby()` <br>

`df.groupby('columnname')` groups rows together according to the values in a column. <br>

`.groupby()` is usually combined with another method that calculates something for each group. <br>

For example: `df.groupby('condition')['duration'].mean()` calculates the average duration for each condition. <br>

You can also group by multiple columns: `df.groupby(['condition', 'correct'])['duration'].mean()` <br>

This gives the average duration for each combination of `condition` and `correct`. <br>

### Calculating several statistics with `.agg()` <br>

`.agg()` stands for aggregate and allows you to calculate several statistics at once. <br>

For example: `df['duration'].agg(['mean', 'min', 'max'])` calculates the mean, minimum, and maximum of `duration`. <br>

`.groupby()` and `.agg()` are especially useful together. <br>

`df.groupby('condition')['duration'].agg(['mean', 'min', 'max'])` calculates the mean, minimum, and maximum duration for each condition. <br>

You can also include `count`: `df.groupby('condition')['duration'].agg(['mean', 'min', 'max', 'count'])` <br>

### Other useful methods <br>

`df['condition'].value_counts()` counts how often each value occurs in a column. <br>

`df['condition'].unique()` returns the unique values in a column. <br>

`df['duration'].mean()` calculates the average. <br>

`df['duration'].median()` calculates the median. <br>

`df['duration'].min()` returns the minimum value. <br>

`df['duration'].max()` returns the maximum value. <br>

`df.sort_values('duration')` sorts the dataframe from the smallest to the largest `duration`. <br>

`df.sort_values('duration', ascending=False)` sorts the dataframe from the largest to the smallest `duration`. <br>
