## Strings

Strings are text, like a word. They are defined by quotation marks, either double or single. These marks are there both in the input and the output. If you put a number in quotation marks, it stops being the actual value of two, and is rather the character depicting two. Thus, a string and an integer or float cannot be added together or used other operations in that manner.

```python
fruit = "2"
fruit
```

To turn the string "2" into an integer, change the type as such:

```python
fruit = int(fruit)
fruit
```

This cannot be done with words in strings, only with numbers.

```python
fruitbowl = "grapes"
fruitbowl = int(fruitbowl)
fruitbowl
#this won't work
```

Strings can be added together as such:

```python
fruit1 = "banana"
fruit2 = "apple"
fruit1 + fruit2
```

Properties of a string. A bit more intuitive than integer properties. The properties show what can be done with the variable, like upper for upper case. Upper and lower case are recognised as two different items; the and The are not the same to python. Therefore it can be useful to change ones data to all lowercase for counting specific words/tokens. 

```python
dir(fruit1)
```

```python
fruit1.upper()
```

```python
fruit1.lower()
```

The replace property asks first for a character in the string and then what you want it to be replaced by.

```python
fruit1.replace('a','@')
```

## Integers

Integers are whole numbers. **No** decimals! Integers can be assigned to variables using =. The variable can't be an integer, like 1=2, but otherwise variables can be named most things. This type of variable is called an integer variable. The variable name can however include a number, like fruit1. 

```python
chips = 2
cookies = 3
snacks = 2 + 3
```

You cannot say chips + cookies = snacks to assign value to snacks, this will cause an error. The name of the variable which you want to assign value to has to be first in the line. You can see the value of a variable by running its name by itself (Jupyter only) or by running print(name). The print command can also be used to see the value of several variables. You can also see the type of variable by running type(name), here being int for integer.

```python
print(chips, cookies, snacks)
```

```python
type(snacks)
```

Everything in Python is an object, and they have different properties. All these variables have different properties. To find them dir(name). Ethan says to ignore all with _property _
