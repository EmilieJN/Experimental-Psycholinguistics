```python
#len
list = ['banana', 'grape', 'apple']
len(list)
```

```python
#slicing
list[1]
```

```python
list[0:2] #from item 0 to item 1
```

```python
list[1:3] #from item 1 to item 2
```

```python
list[::2] #every second item in the entire list, starting with item 0
```

```python
list[1:] #from item 1 to the end of the list
```

```python
list[-1] #the last item in the list
```

```python
list[-2:] #from the second to last item to the end of the list
```

```python
#the first argument represents where for Python to start; second argument shows where to end; 
#third argument shows how many jumps to take
#if the first position is empty, Python assumes the start of the list
#if the second position is empty, Python assumes the end of the list
#if the third position is empty, Python assumes one jump/step
```
