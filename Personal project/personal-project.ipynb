For my personal project, I have coded a short quiz. ## 1. Defining the answers

The questions are yes/no questions, but these variables account for a few different phrasings.

```python
positive_answers = ['yes','Yes', 'y', 'True','true']
negative_answers = ['no','No','n', 'False', 'false']
```

## 2. Defining the questions

I have turned each question/answer combination into a function in order to count the feedback from each question, so a score can be given in the end.

```python
def q1():
    #making the question into an easily manipulable  string variable
    qp= 'Do penguins have knees?'
    #printing the question so it is visible for the quiz-taker
    print(qp)
    #defining the answer as the input from the quiz-taker
    answer = input('> ')
    #preparing feedback for both the quiz-taker and the code 
    #the printed 'Correct!'/'Incorrect!' will be visible for the quiz-taker
    #the returned boolean is for a later function
    if answer in positive_answers:
        print('Correct!')
        return True
    else:
        print('Incorrect!')
        return False
```

I used the _return_ statement in order to get some countable feedback. Initially, I tried to have it return _Correct_ and _Incorrect_ to me, but eventually found it easier to have it return the booleans that it already understood. The following two questions follow the same pattern, except the last question, where the correct answer is amongst _negative_answers_.

```python
def q2():
    qu= 'Is the unicorn the national animal of Scotland?'
    print(qu)
    answer = input('> ')
    if answer in positive_answers:
        print('Correct!')
        return True
    else:
        print('Incorrect!')
        return False
```

```python
def q3():
    qz= 'Is the hybrid of a zebra and a horse called a hobra?'
    print(qz)
    answer = input('> ')
    if answer in negative_answers:
        print('Correct!')
        return True
    else:
        print('Incorrect!')
        return False
```

## 3 Counting the score

The last function runs through each of the questions, counts whether the quiz-taker answers correctly and provides them with a final score of how they did. 

```python
def quiz():
    #the question functions are named variables in order to create a list
    score1 = q1()
    score2 = q2()
    score3 = q3()
    list=[score1,score2,score3]
    #the list contains the returned values of the question functions
    #the .count() method counts the number of True in the list.
    #it prints the quiz-taker's score out of 3
    print('Good job, your final score is', list.count(True), '/ 3!')
```

```python
#running the final function, which runs the quiz in its entirety with feedback
quiz()
```
